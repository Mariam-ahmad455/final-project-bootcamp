```python
df.fillna(0,inplace=True)
```

```python
df['review_score'].fillna(df['review_score'].mean(),inplace=True)
```

```python
df['gender'] = df['gender'].replace({'F': 0, 'M': 1})
df['gender'].fillna(df['gender'].mean(), inplace=True)
```

```python
total_price_array = np.array(df['quantity'] * df['price'])
print(total_price_array)
```

```python
category_name = ['Books & Stationery', 'Electronics', 'Sports & Outdoors']
prices = [20, 50, 100]
categories = []
for i in range(len(category_name)):
    price = prices[i]
    if price < 50:
        categories.append("Affordable")
    elif price >= 50 and price < 100:
        categories.append("Medium Range")
    else:
        categories.append("Premium")
print(categories)
```

```python
plt.figure(figsize=(12, 8))
correlation_matrix =df.select_dtypes(include=['float64', 'int64']).corr()
sns.heatmap(correlation_matrix, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x=df['price'], color='green')
plt.title('Boxplot of the prices')
plt.xlabel('price')
plt.show()
```

```python
plt.hist(total_price_array, bins=20, color='skyblue')
plt.title('Distribution of Total Prices')
plt.xlabel('Total Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(15, 10))
plt.scatter(df['quantity'], df['price'], alpha=0.5, color='blue')
plt.title(' Quantity vs Unit Price')
plt.xlabel('Quantity')
plt.ylabel(' Price')
plt.show()
```
